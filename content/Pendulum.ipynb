(pendulum)=
$$ \ddot{\theta} + \omega^2 \sin(\theta) = 0$$

where $\omega = \sqrt{\frac{g}{L}}$ and $g$ the gravitational acceleration and $L$ the length of rod of the pendulum. 

Analitical solutions are possible with the small angle approximation. Taking a first order taylor expansion of the second term in {numref}`eq {number} <pendulum>` gives the small angle approximation, such as in {numref}`eq {number} <pendulumSmallAngle>`

(pendulumSmallAngle)=
$$ \ddot{\theta} + \omega^2 \theta = 0$$




```python
import numpy as np
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, CustomJS, Slider, Button
from bokeh.layouts import column, row


def eulerCromer(theta0,omega0,L =1 ,g=9.81,A = 0.1,N=1000,tf=10):
    t = np.linspace(0,tf,N)
    dt = t[1]-t[0]
    ws = g/L
    theta = np.zeros(N)
    omega = np.zeros(N)

    #initial conditions
    theta[0] = theta0
    omega[0] = omega0

    for i in range(N-1):
        omega[i+1] = omega[i] - (A*omega[i]*np.abs(omega[i]) + ws*np.sin(theta[i]))*dt 
        theta[i+1] = theta[i] + omega[i+1]*dt 

    x = L*np.sin(theta)
    y = -L*np.cos(theta)
    return t,x,y,theta,omega


t,x,y,theta,omega = eulerCromer(1.1*np.pi,0,L = 1,A = 0.3)

x = np.asarray(x).ravel().tolist()
y = np.asarray(y).ravel().tolist()
t = np.asarray(t).ravel().tolist()


#things
rod = ColumnDataSource(data={"x": [0,x[0]],"y": [0,y[0]]})

bob = ColumnDataSource(data={"x": [x[0]], "y": [y[0]]})

#figure
fig = figure(
    width = 500,
    height = 500,
    x_range = (-1.2,1.2),
    y_range = (-1.2,1.2),
    title = "pendulum title"
)


```
