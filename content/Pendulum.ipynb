(pendulum)=
$$ \ddot{\theta} + \omega^2 \sin(\theta) = 0$$

where $\omega = \sqrt{\frac{g}{L}}$ and $g$ the gravitational acceleration and $L$ the length of rod of the pendulum. 

Analitical solutions are possible with the small angle approximation. Taking a first order taylor expansion of the second term in {numref}`eq {number} <pendulum>` gives the small angle approximation, such as in {numref}`eq {number} <pendulumSmallAngle>`

(pendulumSmallAngle)=
$$ \ddot{\theta} + \omega^2 \theta = 0$$

Where the solution is generally (after initial conditions) is given by eq {numref}`eq {number} <AnalSolutionTheta>`. $\theta_0$ and $\omega_0$ are the initial conditions for the angle $\theta$ and the angular velocity $\omega$ respectivaly. 

(AnalSolutionTheta)=
$$ \theta(t) = \theta_0\cos(\omega t) + \frac{\omega_o}{\omega}\sin(\omega t)$$

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact,FloatSlider






def eulerCromer(theta0,omega0,L =1 ,g=9.81,A = 0.1,N=1000,tf=10):
    t = np.linspace(0,tf,N)
    dt = t[1]-t[0]
    ws = g/L
    theta = np.zeros(N)
    omega = np.zeros(N)

    #initial conditions
    theta[0] = theta0
    omega[0] = omega0

    for i in range(N-1):
        omega[i+1] = omega[i] - (A*omega[i]*np.abs(omega[i]) + ws*np.sin(theta[i]))*dt 
        theta[i+1] = theta[i] + omega[i+1]*dt 

    x = L*np.sin(theta)
    y = -L*np.cos(theta)
    return t,x,y,theta,omega


def Harmonic(theta0,omega0,L,g,N = 1000,tf = 10):
    t = np.linspace(0,tf,N)
    w = np.sqrt(g/L)
    theta = theta0*np.cos(w*t) + omega0/w*np.sin(w*t)
    return t,theta



L = 1
g = 9.81
w = np.sqrt(g/L)
theta0 = 0.1*np.pi
omega0 = 0
time = 20

def plotp(theta0):
    t,thetaH = Harmonic(theta0,omega0,L,g,tf = 20)

    t,x,y,theta,omega = eulerCromer(theta0,omega0,L = 1,A = 0,tf = 20)


    plt.figure()
    plt.plot(t,theta,'k-',label = "numerical solution")
    plt.plot(t,thetaH,'r-', label = "analytical solution")
    plt.legend()
    plt.xlabel("time [s]")
    plt.ylabel("theta [-]")
    plt.show()

interact(
    plotp,
    theta0 = FloatSlider(
        value = 0.1*np.pi,
        min = 0.01,
        max = 0.99*np.pi,
        step = 0.01,
        description = "theta0"
    )
)
```
